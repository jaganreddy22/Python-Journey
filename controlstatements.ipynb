###if statements programs

```python
#special symbol or not
char=input('enter the char: ')
if not ('0'<=char<='9' or 'A'<=char<='Z' or 'a'<=char<='z'):
    print(f'{char} it is a special symbol')
```

```python
#given number is divisible by 3 and 5
n=int(input())
if n%3==0 and n%5==0:
    print(f'{n} is divisble by 3 and 5')
```

```python
#give word is key word
import keyword
s=input()
if s in keyword.kwlist:
    print(f'{s} is key word')
```

```python
#give number is divisble by 2 and 6 if yes convert to complex
n=int(input("enter a number "))
if n%2==0 and n%6==0:
    print(f'{n} is divisible 2 and 6 so it covert into {complex(n)}')
```

```python
#to check last digit end with 5 store in list
n=int(input('Enter a number: '))
arr=[]
last=n%10
if last==5:
    print(f'{n} consist end letter {last}')
    arr.append(last)
    print(arr)
```

```python
#given number is odd store it in tuple
n=int(input())
if n%2!=0:
    res=(n,)
    print(f'{n} is odd so it store in tuple as {res}')
```

```python
#write a program to check mid element if yes print the mid element
word=input('enter a word: ')
if len(word)%2!=0:
    print(word[len(word)//2])
```

```python
#write a program to check mid element if yes print the mid element to word last
word=input('enter a word: ')
if len(word)%2!=0:
    mid=len(word)//2
    print('original name is ',word)
    print('after mid element coverted to last is',word[:mid]+word[mid+1:]+word[mid])
```
